```python
from ultralytics import YOLO
```

```python
model = YOLO("yolov8n.pt")
```

```python
results  = model.train(
    # data =f"{dataset.location}/data.yaml"
    epochs =50,
    imgsz =640,
    batch =16,
    name ="ppe_detection"
)
```

```python
metrics = model.val()
print (metrics)
```
